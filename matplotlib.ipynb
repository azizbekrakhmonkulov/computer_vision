# Multple lines in Numpy and matplotlib train

```python
x = np.array([1, 2, 3, 4])

y1 = np.array([3, 8, 1, 10])
y2 = np.array([5, 2, 9, 6])
y3 = np.array([2, 6, 4, 12])

plt.plot(x, y1, color="green", marker="o", label="Line 1")
plt.plot(x, y2, color="red", linestyle="dotted", marker="s", label="Line 2")
plt.plot(x, y3, color="blue", linestyle="--", marker="^", label="Line 3")

plt.title("Multiple Lines Chart")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.legend()
plt.grid(True)
plt.show()
```

# **Bar Chart with Numpy and Matplotlib**

```python
products = ["Protein", "Creatine", "BCAA", "Vitamin"]
sales = np.array([120, 85, 60, 150])

plt.bar(products, sales, color="blue")

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

```python
months = ["January", "February", "March", "April"]

protein = np.array([120, 150, 180, 210])
creatine = np.array([90, 110, 140, 170])
vitamins = np.array([70, 100, 120, 150])

x = np.arange(len(months))
width = 0.25

# Chart yaratish
fig, ax = plt.subplots(figsize=(11, 6))

bar1 = ax.bar(x - width, protein, width,
              label="Protein", color="#2563eb")

bar2 = ax.bar(x, creatine, width,
              label="Creatine", color="#10b981")

bar3 = ax.bar(x + width, vitamins, width,
              label="Vitamins", color="#f59e0b")

# Ustunlar ustiga qiymat yozish
ax.bar_label(bar1, padding=3)
ax.bar_label(bar2, padding=3)
ax.bar_label(bar3, padding=3)

# Dizayn
ax.set_title("Fitness Store Sales Report",
             fontsize=16, fontweight="bold")

ax.set_xlabel("Months")
ax.set_ylabel("Units Sold")

ax.set_xticks(x)
ax.set_xticklabels(months)

ax.legend()
ax.grid(axis="y", linestyle="--", alpha=0.3)
ax.set_axisbelow(True)
ax.set_ylim(0, 250)

plt.tight_layout()
plt.show()
```

# ***Scatter Plot with Mumpy and Matplotlib***

```python
hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
scores = np.array([35, 45, 50, 65, 70, 80, 85, 95])

plt.scatter(hours, scores, color="red", s=100)

plt.title("Study Hours vs Exam Scores")
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.grid(True, alpha=0.3)

plt.show()
```

```python
np.random.seed(42)

# 50 ta tasodifiy ma'lumot
x = np.random.randint(1, 11, 50)
y = x * 8 + np.random.randint(-15, 16, 50)

sizes = np.random.randint(50, 500, 50)
colors = np.random.rand(50)

plt.figure(figsize=(10, 6))

scatter = plt.scatter(
    x, y,
    s=sizes,
    c=colors,
    cmap="viridis",
    alpha=0.7,
    edgecolors="black"
)

plt.colorbar(scatter, label="Color Scale")

plt.title("Advanced Scatter Plot", fontsize=16)
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.grid(True, linestyle="--", alpha=0.3)

plt.show()
```

# **Histogram with Mumpy and Matplotlib**

```python
scores = np.array([
    45, 55, 60, 65, 70, 72, 75, 78,
    80, 82, 85, 88, 90, 92, 95,
    50, 60, 70, 75, 85
])

plt.hist(scores, bins=5, color="royalblue",
         edgecolor="black")

plt.title("Exam Scores Distribution")
plt.xlabel("Exam Scores")
plt.ylabel("Number of Students")

plt.show()
```

```python
np.random.seed(42)

# 1000 ta talabaning ballari
scores = np.random.normal(70, 15, 1000)
scores = np.clip(scores, 0, 100)

mean_score = np.mean(scores)
median_score = np.median(scores)

plt.figure(figsize=(11, 6))

plt.hist(
    scores,
    bins=20,
    color="royalblue",
    edgecolor="white",
    alpha=0.8
)

# O'rtacha qiymat
plt.axvline(
    mean_score,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Mean: {mean_score:.1f}"
)

# Median qiymat
plt.axvline(
    median_score,
    color="green",
    linestyle=":",
    linewidth=2,
    label=f"Median: {median_score:.1f}"
)

plt.title("Student Exam Score Distribution",
          fontsize=16, fontweight="bold")

plt.xlabel("Exam Scores")
plt.ylabel("Frequency")
plt.legend()
plt.grid(axis="y", alpha=0.2)

plt.tight_layout()
plt.show()
```

# **Subplots**

```python
np.random.seed(42)

x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 25, 15, 35, 45])

fig, ax = plt.subplots(2, 2, figsize=(12, 8))

# 1. Line Chart
ax[0, 0].plot(x, y, color="blue", marker="o")
ax[0, 0].set_title("Line Chart")

# 2. Bar Chart
ax[0, 1].bar(x, y, color="green")
ax[0, 1].set_title("Bar Chart")

# 3. Scatter Plot
ax[1, 0].scatter(x, y, color="orange", s=100)
ax[1, 0].set_title("Scatter Plot")

# 4. Histogram
data = np.random.normal(50, 15, 1000)
ax[1, 1].hist(data, bins=20, color="gray")
ax[1, 1].set_title("Histogram")

# Barcha grafiklar uchun grid
for a in ax.flat:
    a.grid(alpha=0.2)

plt.tight_layout()
fig.savefig("my_graph.png", dpi=300, bbox_inches="tight")
plt.show()
```

# Wireframes with Numpy and Matplotlib

```python
# Ma'lumotlar
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)

X, Y = np.meshgrid(x, y)

# Matematik funksiya
Z = np.sin(np.sqrt(X**2 + Y**2))

# 3D grafik
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

ax.plot_wireframe(
    X, Y, Z,
    color="red",
    linewidth=0.3
)

ax.set_title("3D Wireframe Plot")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")

plt.show()
```
