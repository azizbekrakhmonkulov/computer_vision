```python
import cv2
from google.colab.patches import cv2_imshow

image = cv2.imread("/content/cat_image.jpg")
cv2_imshow(image)
```

```python
print("SHAPE:", image.shape)
print("HEIGHT:", image.shape[0])
print("WIDTH:", image.shape[1])
print("CHANNELS:", image.shape[2])
```

```python
(b, g, r) = image[100, 50]
print("BLUE:", b, "GREEN:", g, "RED:", r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized = cv2.resize(gray, (200, 150))
cv2_imshow(resized)
```

```python
cropped = gray[100:350, 150:400]
cv2_imshow(cropped)
#
```

```python
rotated = cv2.rotate(cropped, cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated)
```

```python
flipped = cv2.flip(cropped, 1)
cv2_imshow(flipped)
#
```

```python
cv2.imwrite("cropped-cat.jpg", cropped)
```

# **Opencv Video streaming codes**

```python
import cv2

face_cascade = cv2.CascadeClassifier(
    "/Users/new/Desktop/opencv/haarcascade_frontalface_default.xml"
)

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Kamera ochilmadi")
    exit()

while True:
    success, frame = cap.read()
    if not success:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, 1.3, 5)

    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

    cv2.imshow("Camera", frame)

    if cv2.waitKey(1) & 0xFF in (ord('q'), 27):
        break

cap.release()
cv2.destroyAllWindows()
# ========== OpenCV Video Capture Example ==========
# import cv2

# cap = cv2.VideoCapture(0)

# while True:
#     sucsess, frame = cap.read()
#     if not sucsess:
#         break

#     gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
#     blurred = cv2.GaussianBlur(frame, (5, 5), 0)
#     # edges = cv2.Canny(frame, 100, 200)
#     flipped = cv2.flip(frame, 1)

#     # cv2.imshow("Original", frame)
#     # cv2.imshow("Gray", gray)
#     # cv2.imshow("Blurred", blurred)
#     # cv2.imshow("Edges", edges)
#     cv2.imshow("Flipped", flipped)

#     if cv2.waitKey(1) & 0xFF == ord('q'):
#         break

# cap.release()
# cv2.destroyAllWindows()
```
